```python
import pandas as pd
import numpy as np

# Load dataset (Windows path fixed)
df = pd.read_csv(r"C:\Users\vinay\Downloads\netflix_titles.csv\netflix_titles.csv") 

# Check initial shape and first 5 rows
print("Initial data shape:", df.shape)
df.head()
```

```python
# Check which columns have missing values
print(df.isnull().sum())
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df.info()
df.head()
```

```python
# Checking duplicates
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
# Lowercase or uppercase for consistency
df['type'] = df['type'].str.lower()
df['rating'] = df['rating'].str.upper() 
df['country'] = df['country'].str.title()
df.head()
```

```python
# Converting 'date_added' to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')  
df['date_added'].head()
df.head()
```

```python
# Rename columns to lowercase and removing spaces
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.head()
```

```python
# Converting release_year to integer
df['release_year'] = df['release_year'].astype(int)
df.dtypes
```

```python
# Remove unreasonable release years
df = df[df['release_year'] >= 1900]
```

```python
df.to_csv("cleaned_netflix_titles.csv", index=False)
print("Cleaned dataset saved as 'cleaned_netflix_titles.csv'")
```

```python
summary = """
Data Cleaning Summary:
- Removed duplicate rows
- Filled missing values (director, cast, country)
- Standardized text columns (type, rating, country)
- Converted 'date_added' column to datetime format
- Renamed column headers to lowercase with underscores
- Converted 'release_year' to integer
- Removed outliers with unrealistic release years
"""
with open("README.md", "w") as f:
    f.write(summary)
print("README.md created")
```
